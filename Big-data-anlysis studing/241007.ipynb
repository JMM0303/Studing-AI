```python
import pandas as pd

insurance_df = pd.read_csv("D:/insurance_data.csv")

insurance_df.info()
```

```python
titanic_df.columns
```

```python
insurance_df.drop(columns='index', inplace=True)
```

```python
insurance_df.duplicated()
```

```python
insurance_df.dropna()
```

```python
insurance_df.rename(columns={'Gender':'gender'}, inplace=True)
insurance_df.rename(columns={'claim':'charges'}, inplace=True)
```

```python
insurance_df.replace({'gender':{'Male':0}},inplace=True)
insurance_df.replace({'gender':{'Female':1}},inplace=True)

insurance_df.head(10)
```

```python
sns.boxplot(y = insurance_df['charges'] )
```

```python
insurance_df.drop(insurance_df[insurance_df.charges >= 55000].index, inplace = True)
sns.boxplot(y = insurance_df['charges'] )
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
insurance_df[['charges']] = scaler.fit_transform(insurance_df[['charges']])
insurance_df.describe()
```

```python
insurance_df.replace({'diabetic':{'No':0}},inplace=True)
insurance_df.replace({'diabetic':{'Yes':1}},inplace=True)
insurance_df.replace({'smoker':{'No':0}},inplace=True)
insurance_df.replace({'smoker':{'Yes':1}},inplace=True)
insurance_df.replace({'region':{'southeast':0}},inplace=True)
insurance_df.replace({'region':{'northwest':1}},inplace=True)
insurance_df.replace({'region':{'southwest':2}},inplace=True)
insurance_df.replace({'region':{'northeast':2}},inplace=True)

sns.heatmap(insurance_df.corr(), annot=True)
```

```python
insurance_df.drop(columns='age', inplace=True)
insurance_df.drop(columns='gender', inplace=True)
insurance_df.drop(columns='diabetic', inplace=True)

insurance_df.head(10)
```
